```python
#import libraries
import pandas as pd
from datetime import datetime, timedelta
import boto3
from io import BytesIO
```

### Merge Artificial Plants CSV Files Weekly

```python
# Configuration
bucket_name = 'csis4495'
aws_region = 'us-west-2' 

# Initialize a boto3 S3 client
s3_client = boto3.client('s3', region_name=aws_region)

# Define the date range
start_date = datetime(2024, 2, 26)
end_date = datetime(2024, 3, 3)

# Generate a list of date strings that fall within the specified range
date_range = [start_date + timedelta(days=x) for x in range((end_date - start_date).days + 1)]
folder_name = 'daily_artificial_plants'  # Folder name in the S3 bucket
date_strings = [f"{folder_name}/{date.strftime('products_artifical_plants_%Y_%m_%d.csv')}" for date in date_range]


# Read each file into a DataFrame and append it to a list
dataframes = []
for date_string in date_strings:
    file_key = date_string  # Assuming the files are named exactly like the date_strings and are in the root of the bucket
    try:
        # Use boto3 to get the object
        obj = s3_client.get_object(Bucket=bucket_name, Key=file_key)
        # Read the object's content into a DataFrame
        df = pd.read_csv(BytesIO(obj['Body'].read()), encoding='latin1')
        dataframes.append(df)
    except s3_client.exceptions.NoSuchKey:
        print(f"File not found: {file_key}")
    except Exception as e:
        print(f"Error processing file {file_key}: {e}")

# Concatenate all DataFrames into a single DataFrame
combined_df = pd.concat(dataframes, ignore_index=True) if dataframes else pd.DataFrame()

combined_df
# `combined_df` contains all the data from the matching files

# Format the output filename dynamically based on the start_date and end_date
start_date_str = start_date.strftime('%Y_%m_%d')
end_date_str = end_date.strftime('%Y_%m_%d')
output_filename = f'weekly_products_artificial_plants_{start_date_str}_to_{end_date_str}.csv'

# Save the combined DataFrame to a new CSV file
combined_df.to_csv(output_filename, index=False, encoding='latin1') 

print(f"CSV file has been created : '{output_filename}'.")

AWS_ACCESS_KEY='AKIAZQ3DRYKPWCEBEPGF'
AWS_SECRET_KEY='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'
AWS_S3_BUCKET_NAME ='csis4495'
AWS_REGION="us-west-2"

LOCAL_FILE = output_filename
NAME_FOR_S3 = f'weekly_artificial_plants/{output_filename}'
def main():
    print('in main method')

    s3_client = boto3.client(
        service_name='s3',
        region_name=AWS_REGION,
        aws_access_key_id = AWS_ACCESS_KEY,
        aws_secret_access_key = AWS_SECRET_KEY
    )
    response = s3_client.upload_file(LOCAL_FILE, AWS_S3_BUCKET_NAME, NAME_FOR_S3)

    print(f'upload file response : Succesfully Uploaded to AWS S3')

if __name__ == '__main__':
    main()
```
